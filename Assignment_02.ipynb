#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



#1

### Construction of the Probability Space $(\Omega, \mathcal{F}, P)$

**1. The Sample Space ($\Omega$):**
Since the random variable $X$ directly tracks the specific system states, we define the sample space $\Omega$ as the finite set of all possible atomic outcomes:
$$\Omega = \{0, 1, 2, 3\}$$

In this setup, each distinct outcome $\omega \in \Omega$ directly represents the exact number of emergency transformer failures reported within a single operational day.

**2. The Sigma-Algebra ($\mathcal{F}$):**
Because the underlying sample space $\Omega$ is finite, we can safely and naturally utilize the complete power set of $\Omega$ to construct our collection of measurable events:
$$\mathcal{F} = \mathcal{P}(\Omega)$$

Explicitly expanding this power set provides a robust $\sigma$-algebra containing all $2^4 = 16$ possible experimental configurations:
$$\mathcal{F} = \{ \emptyset, \{0\}, \{1\}, \{2\}, \{3\}, \{0,1\}, \{0,2\}, \{0,3\}, \{1,2\}, \{1,3\}, \{2,3\}, \{0,1,2\}, \{0,1,3\}, \{0,2,3\}, \{1,2,3\}, \Omega \}$$

**3. The Probability Measure ($P$):**
The probability measure $P : \mathcal{F} \to [0,1]$ is formally established by mapping the given operational likelihoods directly to each elementary singleton event:
$$P(\{0\}) = 0.50$$
$$P(\{1\}) = 0.30$$
$$P(\{2\}) = 0.15$$
$$P(\{3\}) = 0.05$$

By applying the fundamental axiom of finite additivity for disjoint sets, the total probability of any arbitrary composite event $A \in \mathcal{F}$ is computed systematically by summing the weights of its individual elements:
$$\boxed{P(A) = \sum_{\omega \in A} P(\{\omega\})}$$

#2

### Formal Definition of the Random Variable $X$


A random variable on a probability space $(\Omega, \mathcal{F}, P)$ is defined as a measurable function mapping elements from the sample space to the real numbers. We define the random variable $X$ by:
$$X : \Omega \to \mathbb{R}$$

under the explicit rule:
$$X(\omega) = \omega, \quad \forall \omega \in \Omega$$


Since the underlying sample space consists of the finite atomic states $\Omega = \{0, 1, 2, 3\}$, evaluating this functional rule across all individual sample points maps them as:
$$X(0) = 0, \quad X(1) = 1, \quad X(2) = 2, \quad X(3) = 3$$

 $X$ simply returns the exact count of emergency transformer failures recorded during a single operational day.

### Probability Mass Function of $X$

The probability mass function (PMF) of a discrete random variable $X$ gives the probability assigned to each possible value of $X$. It is defined as
$$
p_X(x)=P(X=x)
$$

Since the random variable $X$ represents the number of emergency transformer failures in a single day, it takes values in the finite set
$$
\{0,1,2,3\}.
$$

Using the given probabilities, the PMF of $X$ is

$$
p_X(0)=P(X=0)=0.50
$$

$$
p_X(1)=P(X=1)=0.30
$$

$$
p_X(2)=P(X=2)=0.15
$$

$$
p_X(3)=P(X=3)=0.05
$$

Therefore, the probability mass function can be written as

$$
\boxed{
p_X(x)=
\begin{cases}
0.50, & x=0,\\
0.30, & x=1,\\
0.15, & x=2,\\
0.05, & x=3,\\
0, & \text{otherwise}.
\end{cases}
}
$$

To verify that this is a valid PMF, we check the normalization condition:
$$
\sum_{x\in\{0,1,2,3\}}p_X(x)
=
0.50+0.30+0.15+0.05
=
1.
$$

Hence, $p_X$ is a valid probability mass function.

### Expectation of $X$

The expectation of a discrete random variable $X$ is defined as

$$
E[X]=\sum_x xP(X=x).
$$

Since $X$ takes values in $\{0,1,2,3\}$, we have

$$
E[X]
=
0.P(X=0)+1.P(X=1)+2.P(X=2)+3.P(X=3).
$$

Substituting the given probabilities,

$$
E[X]
=
0.(0.50)+1.(0.30)+2.(0.15)+3.(0.05).
$$

Thus,

$$
E[X]
=
0+0.30+0.30+0.15
=
0.75.
$$

Therefore,

$$
\boxed{E[X]=0.75}.
$$

Hence, the expected number of emergency transformer failures reported in a single day is $0.75$.

```python
# Values of X
x_values = [0, 1, 2, 3]

# Corresponding probabilities
probabilities = [0.50, 0.30, 0.15, 0.05]

# Expectation E[X] = sum x * P(X=x)
expected_value = sum(x * p for x, p in zip(x_values, probabilities))

print("E[X] =", expected_value)
```

### Variance of $X$

The variance of a discrete random variable $X$ is defined as

$$
\operatorname{Var}(X)=E[X^2]-(E[X])^2.
$$

From the previous calculation,

$$
E[X]=0.75.
$$

Now,

$$
E[X^2]=\sum_x x^2P(X=x).
$$

Since $X$ takes values in $\{0,1,2,3\}$,

$$
E[X^2]
=
0^2.(0.50)+1^2.(0.30)+2^.2.(0.15)+3^2.(0.05).
$$

Thus,

$$
E[X^2]
=
0+0.30+4(0.15)+9(0.05).
$$

Therefore,

$$
E[X^2]
=
0.30+0.60+0.45
=
1.35.
$$

Hence,

$$
\operatorname{Var}(X)
=
1.35-(0.75)^2.
$$

So,

$$
\operatorname{Var}(X)
=
1.35-0.5625
=
0.7875.
$$

Therefore,

$$
\boxed{\operatorname{Var}(X)=0.7875}.
$$

```python
import numpy as np

x_values = np.array([0, 1, 2, 3])
probabilities = np.array([0.50, 0.30, 0.15, 0.05])

expected_value = np.sum(x_values * probabilities)
expected_square = np.sum((x_values**2) * probabilities)

variance = expected_square - expected_value**2

print("E[X] =", expected_value)
print("E[X^2] =", expected_square)
print("Var(X) =", variance)
```

#1

### Construction of the Probability Space $(\Omega,\mathcal{F},P)$

The random variable \(X\) represents the proportion of defective items in a large batch. Since \(X\) is a proportion, it takes values in the interval \([0,1]\). Therefore, we define the  **sample space** as

$$
\Omega=[0,1].
$$

Each outcome \(ω∈Ω\) represents a possible defective proportion in the batch.

Since the sample space is a continuous interval, we take the sigma-algebra to be the Borel sigma-algebra on \([0,1]\):

$$
\mathcal{F}=\mathcal{B}([0,1]).
$$

The probability measure \(P\) is induced by the Beta distribution with parameters \(α=2\) and \(β=18\). The density function is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\
0, & \text{otherwise}.
\end{cases}
$$

Thus, for any event A, the probability measure is defined by

$$
P(A)=\int_A f(x)\,dx.
$$

Equivalently,

$$
P(A)=
\int_A
\dfrac{1}{B(2,18)}
x(1-x)^{17}
\,dx.
$$

Therefore, a suitable probability space is

$$
\boxed{
(\Omega,\mathcal{F},P)
=
\left(
[0,1],
\mathcal{B}([0,1]),
P
\right)
}.
$$

### Probability Density Function of $X$

Since $X$ follows a Beta distribution with parameters $\alpha=2$ and $\beta=18$, its probability density function is

$$
f_X(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{2-1}(1-x)^{18-1}, & 0<x<1,\\[6pt]

Simplifying the powers gives

f_X(x)=

### Verification that $f$ is a Valid Probability Density Function

To verify that $f$ is a valid probability density function, we must check two conditions:

$$
f(x)\geq 0
$$

for all $x$, and

$$
\int_{-\infty}^{\infty} f(x)\,dx=1.
$$

The density is given by

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\[6pt]
0, & \text{otherwise}.
\end{cases}
$$

First, for $0<x<1$, we have

$$
x>0
$$

and

$$
(1-x)^{17}>0.
$$

Also, since the Beta function is positive,

$$
B(2,18)>0.
$$

Therefore,

$$
f(x)\geq 0.
$$

Outside the interval $(0,1)$, we have

$$
f(x)=0.
$$

Hence, $f$ is non-negative everywhere.

Next, we verify that the total integral is equal to $1$:

$$
\int_{-\infty}^{\infty} f(x)\,dx
=
\int_0^1 \frac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Taking the constant outside,

$$
\int_{-\infty}^{\infty} f(x)\,dx
=
\frac{1}{B(2,18)}
\int_0^1 x(1-x)^{17}\,dx.
$$

By the definition of the Beta function,

$$
B(2,18)
=
\int_0^1 x^{2-1}(1-x)^{18-1}\,dx.
$$

Thus,

$$
B(2,18)
=
\int_0^1 x(1-x)^{17}\,dx.
$$

Therefore,

$$
\int_{-\infty}^{\infty} f(x)\,dx
=
\frac{1}{B(2,18)}B(2,18)
=
1.
$$

Since $f(x)\geq 0$ for all $x$ and its total integral is $1$, $f$ is a valid probability density function.

$$
\boxed{f \text{ is a valid probability density function.}}
$$

### Expectation of $X$

Since $X$ follows a Beta distribution with parameters $\alpha=2$ and $\beta=18$, we use the expectation formula for a Beta random variable:

$$
E[X]=\frac{\alpha}{\alpha+\beta}.
$$

Substituting $\alpha=2$ and $\beta=18$,

$$
E[X]=\frac{2}{2+18}.
$$

Thus,

$$
E[X]=\frac{2}{20}=0.10.
$$

Therefore,

$$
\boxed{E[X]=0.10}.
$$

Hence, the expected defective proportion in the batch is $0.10$, or $10\%$.

### Variance of $X$

Since $X$ follows a Beta distribution with parameters $\alpha=2$ and $\beta=18$, the variance of a Beta random variable is given by

$$
\operatorname{Var}(X)
=
\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}.
$$

Substituting $\alpha=2$ and $\beta=18$,

$$
\operatorname{Var}(X)
=
\frac{(2)(18)}{(2+18)^2(2+18+1)}.
$$

Thus,

$$
\operatorname{Var}(X)
=
\frac{36}{20^2(21)}.
$$

Therefore,

$$
\operatorname{Var}(X)
=
\frac{36}{8400}
=
0.0042857.
$$

Hence,

$$
\boxed{\operatorname{Var}(X)\approx 0.00429}.
$$

This means that the defective proportion varies around its mean value of $0.10$, but the variation is relatively small.

```python
# Beta distribution parameters
alpha = 2
beta = 18

# Expectation
expected_value = alpha / (alpha + beta)

# Variance
variance = (alpha * beta) / ((alpha + beta)**2 * (alpha + beta + 1))

print("E[X] =", expected_value)
print("Var(X) =", variance)
```

### Probability that the defective proportion is at most $0.10$

Since $X\sim \text{Beta}(2,18)$, its density is

$$
f_X(x)=\frac{1}{B(2,18)}x(1-x)^{17},\qquad 0<x<1.
$$

Therefore,

$$
P(X\leq 0.10)
=
\int_0^{0.10} f_X(x)\,dx.
$$

Substituting the density,

$$
P(X\leq 0.10)
=
\int_0^{0.10}
\frac{1}{B(2,18)}
x(1-x)^{17}
\,dx.
$$

Thus,

$$
\boxed{
P(X\leq 0.10)
=
\int_0^{0.10}
\frac{1}{B(2,18)}

```python
from scipy.stats import beta

alpha = 2
beta_param = 18

probability = beta.cdf(0.10, alpha, beta_param)

print("P(X <= 0.10) =", probability)
```

### Practical Reason for Using a Beta Distribution

A Beta distribution is a reasonable model in this situation because the random variable \(X\) represents a proportion of defective items. A proportion must lie between \(0\) and \(1\), and the Beta distribution is naturally defined on the interval \([0,1]\).

The Beta distribution is also flexible. By changing the parameters \(𝛂) and \(β\), it can model different possible shapes for proportions, including low defect rates, high defect rates, and skewed distributions.

In this case,

$$
X\sim \text{Beta}(2,18).
$$

The expected value is

$$
E[X]=\frac{2}{2+18}=0.10.
$$

Thus, the distribution is centered around a defective proportion of approximately \(10\%\). Since \(β=18\) is much larger than \(𝛂=2\), the distribution places more probability near smaller values of \(X\). This is realistic for a manufacturing process where the proportion of defective items is expected to be relatively low.
