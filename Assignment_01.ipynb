#Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius  10 . Let  A  be the event that the point lies within  1  unit of the boundary.

1.Model the experiment assuming the point is uniformly distributed with respect to area, and compute  P(A) .

2.Model the experiment assuming the distance to the point from the center is chosen uniformly from  [0,10] , and the direction is chosen independently and uniformly from  [0,2π)  and compute  P(A) .

3.Explain why the two answers differ.

# **For Part 1**

To solve this, We want to Understand some theories,

1. The Sample Space ($\Omega$): In a continuous space like a disk, $\Omega$ is the
set of all points $(x, y)$ such that their distance from the center $(0,0)$ is less than or equal to the radius $R=10$.

2. The Event ($A$): This is the "boundary region." Since the radius is $10$, being within $1$ unit of the boundary means the point's distance $r$ from the center is between $9$ and $10$.

3. Uniform Distribution (Area): When an experiment is "uniformly distributed with respect to area," the probability of an event is simply the Area of the Event divided by the Total Area of the Sample Space.

### Mathematical Modeling

**1. Sample Space ($\Omega$):**
The set of all points within the disk of radius $R = 10$:
$$\Omega = \{ (x, y) \in \mathbb{R}^2 : \sqrt{x^2 + y^2} \le 10 \}$$

**2. Event ($A$):**
The point lies within 1 unit of the boundary (between radius 9 and 10):
$$A = \{ (x, y) \in \mathbb{R}^2 : 9 \le \sqrt{x^2 + y^2} \le 10 \}$$

**3. Probability Calculation:**
Since the distribution is uniform with respect to area:
$$P(A) = \frac{\text{Area}(A)}{\text{Area}(\Omega)}$$
$$\text{Area}(\Omega) = \pi(10)^2 = 100\pi$$
$$\text{Area}(A) = \pi(10)^2 - \pi(9)^2 = 100\pi - 81\pi = 19\pi$$
$$P(A) = \frac{19\pi}{100\pi} = 0.19$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Create the figure
fig, ax = plt.subplots(figsize=(8, 8))

# Define the disks
theta = np.linspace(0, 2*np.pi, 100)
x_outer = 10 * np.cos(theta)
y_outer = 10 * np.sin(theta)
x_inner = 9 * np.cos(theta)
y_inner = 9 * np.sin(theta)

# Plot the total Sample Space (Omega)
ax.plot(x_outer, y_outer, color='black', label='Boundary (r=10)')
ax.fill(x_outer, y_outer, color='lightgray', alpha=0.3, label='Sample Space $\Omega$')

# Highlight Event A (The boundary region)
ax.fill(np.append(x_outer, x_inner[::-1]),
        np.append(y_outer, y_inner[::-1]),
        color='orange', alpha=0.6, label='Event A (Boundary Region)')

# Formatting
ax.set_aspect('equal')
ax.set_xlim(-11, 11)
ax.set_ylim(-11, 11)
ax.axhline(0, color='black', lw=1)
ax.axvline(0, color='black', lw=1)
ax.set_title("Visualization of Event A in Sample Space $\Omega$")
ax.legend()

plt.grid(True, linestyle='--')
plt.show()

print(f"Calculated Probability P(A): {19/100}")
```

**Python code to compute the probability**

```python
import numpy as np

R_outer = 10
R_inner = 9

area_full_disk = np.pi * R_outer**2
area_inner_disk = np.pi * R_inner**2
area_event_A = area_full_disk - area_inner_disk

probability_A = area_event_A / area_full_disk

print("Area of full disk =", area_full_disk)
print("Area of event A =", area_event_A)
print("P(A) =", probability_A)
```

01.**Uniform Distance vs. Uniform Area**

The key difference here is how the probability measure is defined.

Part 1 (Uniform Area): Points are more likely to be found further from the center because there is more area at larger radius.

Part 2 (Uniform Distance/Direction): Every distance from the center ($0$ to $10$) is equally likely. This means points will appear much more crowded near the center of the disk because the same number of points are squeezed into a smaller area.




### Mathematical Modeling (Model 2: Uniform Distance and Direction)

**1. Definitions:**
Let $R$ be the random variable representing the distance from the center, and $\Theta$ be the random variable representing the direction.
- $R \sim \text{Uniform}(0, 10)$
- $\Theta \sim \text{Uniform}(0, 2\pi)$
- $R$ and $\Theta$ are independent.

**2. Event ($A$):**
The event $A$ occurs if the distance $r$ is within 1 unit of the outer boundary ($r=10$):
$$A = \{ (r, \theta) : 9 \le r \le 10 \}$$

**3. Probability Calculation:**
Because $R$ is uniformly distributed over the interval $[0, 10]$, the probability depends only on the length of the radial interval:
$$P(A) = P(9 \le R \le 10)$$
$$P(A) = \int_{9}^{10} f_R(r) \, dr = \int_{9}^{10} \frac{1}{10} \, dr$$
$$P(A) = \left[ \frac{r}{10} \right]_{9}^{10} = \frac{10}{10} - \frac{9}{10} = 0.1$$

**Conclusion:**
In this model, $P(A) = 0.1$. Note that this is lower than the area-based model ($0.19$) because this model places more probability mass near the center.

# Q2

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

We show that different natural meanings of “random chord” give different probabilities.

### Basic Geometry

Let the circle have radius $R$.

The side length of the inscribed equilateral triangle is:
$$s = \sqrt{3}R$$

For a chord whose perpendicular distance from the center is $d$, the chord length is:
$$L = 2\sqrt{R^2 - d^2}$$

We want the chord length to exceed the triangle's side length:
$$L > \sqrt{3}R$$

Substituting the equation for $L$ into the inequality:
$$2\sqrt{R^2 - d^2} > \sqrt{3}R$$

Squaring both sides:
$$4(R^2 - d^2) > 3R^2$$
$$4R^2 - 4d^2 > 3R^2$$
$$R^2 > 4d^2$$

Dividing by 4:
$$d^2 < \frac{R^2}{4}$$

Taking the square root (since distance $d$ must be positive):
$$d < \frac{R}{2}$$

**Conclusion:**
The chord is longer than the side of the equilateral triangle if and only if its perpendicular distance from the center is less than $\frac{R}{2}$.

### Model 1: Random Endpoints on the Circumference

In this model, we choose two points independently and uniformly on the circle. By rotational symmetry, we can fix one endpoint and choose the other uniformly around the circle. Let $\theta$ be the angle between the two endpoints.

The chord length is given by:
$$L = 2R\sin\left(\frac{\theta}{2}\right)$$

We need the chord length to exceed the triangle's side length ($L > \sqrt{3}R$):
$$2R\sin\left(\frac{\theta}{2}\right) > \sqrt{3}R$$

Dividing both sides by $2R$:
$$\sin\left(\frac{\theta}{2}\right) > \frac{\sqrt{3}}{2}$$

Based on the unit circle, this happens when:
$$\frac{\theta}{2} \in \left(\frac{\pi}{3}, \frac{2\pi}{3}\right)$$

Multiplying by 2 to solve for the full angle $\theta$:
$$\theta \in \left(\frac{2\pi}{3}, \frac{4\pi}{3}\right)$$

The length of this "success" interval is:
$$\frac{4\pi}{3} - \frac{2\pi}{3} = \frac{2\pi}{3}$$

Since the second endpoint is chosen uniformly around the entire circle, $\theta$ is uniformly distributed over $[0, 2\pi)$. The probability is simply the length of our successful interval divided by the total possible interval:
$$P = \frac{\frac{2\pi}{3}}{2\pi}$$

$$P = \frac{1}{3}$$

**Conclusion:**
Therefore, under the random endpoints model, the probability is:
$$\boxed{P = \frac{1}{3}}$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility (optional)
np.random.seed(42)

# Number of random chords to generate
num_chords = 500
R = 1 # Radius of the circle

# 1. Generate random angles for the two endpoints on the circumference
# Both endpoints are chosen uniformly from 0 to 2*pi
theta1 = np.random.uniform(0, 2*np.pi, num_chords)
theta2 = np.random.uniform(0, 2*np.pi, num_chords)

# 2. Convert angles to Cartesian (x, y) coordinates
x1, y1 = R * np.cos(theta1), R * np.sin(theta1)
x2, y2 = R * np.cos(theta2), R * np.sin(theta2)

# 3. Calculate chord lengths using the distance formula
lengths = np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# 4. Check for success (length > side of inscribed equilateral triangle)
# The side of the inscribed triangle is R * sqrt(3)
success = lengths > (np.sqrt(3) * R)

# Calculate empirical probability
prob = np.sum(success) / num_chords

# --- VISUALIZATION ---
fig, ax = plt.subplots(figsize=(8, 8))

# Draw the main outer circle
circle = plt.Circle((0, 0), R, color='black', fill=False, lw=2)
ax.add_patch(circle)

# Draw the inner boundary circle (Radius = R/2)
# A chord is longer than the triangle's side if its midpoint falls inside this circle
inner_circle = plt.Circle((0, 0), R/2, color='red', fill=False, linestyle='--', lw=1.5, label='Inner Radius (R/2)')
ax.add_patch(inner_circle)

# Draw an inscribed equilateral triangle for reference
t_angles = np.array([np.pi/2, np.pi/2 + 2*np.pi/3, np.pi/2 + 4*np.pi/3, np.pi/2])
ax.plot(R * np.cos(t_angles), R * np.sin(t_angles), 'k--', lw=1.5, label='Inscribed Triangle')

# Plot the chords
for i in range(num_chords):
    if success[i]:
        # Successful chords (orange)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='orange', alpha=0.4, lw=1)
    else:
        # Unsuccessful chords (blue)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='blue', alpha=0.15, lw=1)

# Add dummy lines for the legend so we don't have 500 legend entries
ax.plot([], [], color='orange', label=r'Length $> R\sqrt{3}$')
ax.plot([], [], color='blue', label=r'Length $\leq R\sqrt{3}$')

# Formatting the plot
ax.set_aspect('equal')
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.axis('off') # Hide axes for a cleaner look
ax.set_title(f"Bertrand's Paradox: Model 1 (Random Endpoints)\nEmpirical Probability = {prob:.3f} (Theoretical = 0.333)")
ax.legend(loc='upper right')

plt.show()
```

### Model 2: Random Distance from the Center

In this model, we choose the perpendicular distance $d$ of the chord from the center uniformly from the interval:
$$[0, R]$$

Then, we choose the orientation of the chord uniformly.

As established in the basic geometry derivation, the chord is longer than the side of the inscribed equilateral triangle when its distance to the center satisfies:
$$d < \frac{R}{2}$$

Since the distance $d$ is uniformly distributed over $[0, R]$ (denoted as $d \sim U(0, R)$), the probability is simply the length of our "success" interval divided by the length of the total possible interval:
$$P = \frac{\frac{R}{2}}{R}$$

$$P = \frac{1}{2}$$

**Conclusion:**
Therefore, under the random distance model, the probability is:
$$\boxed{P = \frac{1}{2}}$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility (optional)
np.random.seed(42)

# Number of random chords to generate
num_chords = 500
R = 1 # Radius of the circle

# 1. Generate random perpendicular distances (d) from the center
# d is uniformly distributed between 0 and R
d = np.random.uniform(0, R, num_chords)

# Generate a random orientation/angle for each chord
alpha = np.random.uniform(0, 2*np.pi, num_chords)

# 2. Calculate the midpoints of the chords using polar to cartesian
x_m = d * np.cos(alpha)
y_m = d * np.sin(alpha)

# 3. Calculate the half-length of each chord using the Pythagorean theorem
# Since the radius R forms a right triangle with d and half the chord length
L_half = np.sqrt(R**2 - d**2)

# 4. Calculate the endpoints of the chords
# The chord is perpendicular to its distance vector, so we offset the angle by 90 degrees (pi/2)
x1 = x_m - L_half * np.sin(alpha)
y1 = y_m + L_half * np.cos(alpha)

x2 = x_m + L_half * np.sin(alpha)
y2 = y_m - L_half * np.cos(alpha)

# 5. Check for success (distance from center < R/2)
success = d < (R / 2)

# Calculate empirical probability
prob = np.sum(success) / num_chords

# --- VISUALIZATION ---
fig, ax = plt.subplots(figsize=(8, 8))

# Draw the main outer circle
circle = plt.Circle((0, 0), R, color='black', fill=False, lw=2)
ax.add_patch(circle)

# Draw the inner boundary circle (Radius = R/2)
inner_circle = plt.Circle((0, 0), R/2, color='red', fill=False, linestyle='--', lw=1.5, label='Inner Radius (R/2)')
ax.add_patch(inner_circle)

# Draw an inscribed equilateral triangle for reference
t_angles = np.array([np.pi/2, np.pi/2 + 2*np.pi/3, np.pi/2 + 4*np.pi/3, np.pi/2])
ax.plot(R * np.cos(t_angles), R * np.sin(t_angles), 'k--', lw=1.5, label='Inscribed Triangle')

# Plot the chords
for i in range(num_chords):
    if success[i]:
        # Successful chords (orange)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='orange', alpha=0.4, lw=1)
    else:
        # Unsuccessful chords (blue)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='blue', alpha=0.15, lw=1)

# Add dummy lines for the legend
ax.plot([], [], color='orange', label=r'Length $> R\sqrt{3}$')
ax.plot([], [], color='blue', label=r'Length $\leq R\sqrt{3}$')

# Formatting the plot
ax.set_aspect('equal')
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.axis('off')
ax.set_title(f"Bertrand's Paradox: Model 2 (Random Distance)\nEmpirical Probability = {prob:.3f} (Theoretical = 0.500)")
ax.legend(loc='upper right')

plt.show()
```

### Model 3: Random Midpoint Uniformly Chosen in the Disk

In this model, we choose the midpoint of the chord uniformly with respect to area inside the disk. A chord is uniquely determined by its midpoint and orientation.

As established, the chord is longer than the side of the inscribed equilateral triangle when its midpoint lies within a distance of:
$$\frac{R}{2}$$
from the center.

This means the favorable region (where success occurs) is an inner disk of radius $\frac{R}{2}$. The total sample space is the full disk of radius $R$.

Since the midpoint is chosen uniformly with respect to area, the probability is the ratio of these two areas:
$$P = \frac{\text{Area of disk of radius } R/2}{\text{Area of disk of radius } R}$$

Substituting the formula for the area of a circle ($A = \pi r^2$):
$$P = \frac{\pi\left(\frac{R}{2}\right)^2}{\pi R^2}$$

$$P = \frac{\frac{\pi R^2}{4}}{\pi R^2}$$

The $\pi R^2$ terms cancel out, leaving:
$$P = \frac{1}{4}$$

**Conclusion:**
Therefore, under the random midpoint model, the probability is:
$$\boxed{P = \frac{1}{4}}$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

num_chords = 500
R = 1

# 1. Generate random midpoints uniformly by area
# To sample uniformly in a circle, we take the square root of a random uniform variable
# This matches the "Uniform by Area" concept from your first assignment!
u1 = np.random.uniform(0, 1, num_chords)
u2 = np.random.uniform(0, 2*np.pi, num_chords)

r_mid = R * np.sqrt(u1) # Distance from center
theta_mid = u2          # Angle

# Calculate midpoint Cartesian (x, y) coordinates
x_m = r_mid * np.cos(theta_mid)
y_m = r_mid * np.sin(theta_mid)

# 2. Calculate the half-length of each chord using Pythagorean theorem
L_half = np.sqrt(R**2 - r_mid**2)

# 3. Calculate the endpoints of the chords
# The chord is perpendicular to its radial vector (theta_mid), so we add 90 degrees (pi/2)
alpha = theta_mid + np.pi / 2

x1 = x_m - L_half * np.cos(alpha)
y1 = y_m - L_half * np.sin(alpha)

x2 = x_m + L_half * np.cos(alpha)
y2 = y_m + L_half * np.sin(alpha)

# 4. Check for success (midpoint distance from center < R/2)
success = r_mid < (R / 2)

# Calculate empirical probability
prob = np.sum(success) / num_chords

# --- VISUALIZATION ---
fig, ax = plt.subplots(figsize=(8, 8))

# Draw the main outer circle
circle = plt.Circle((0, 0), R, color='black', fill=False, lw=2)
ax.add_patch(circle)

# Draw the inner boundary circle (Radius = R/2)
inner_circle = plt.Circle((0, 0), R/2, color='red', fill=False, linestyle='--', lw=1.5, label='Inner Radius (R/2)')
ax.add_patch(inner_circle)

# Draw an inscribed equilateral triangle for reference
t_angles = np.array([np.pi/2, np.pi/2 + 2*np.pi/3, np.pi/2 + 4*np.pi/3, np.pi/2])
ax.plot(R * np.cos(t_angles), R * np.sin(t_angles), 'k--', lw=1.5, label='Inscribed Triangle')

# Plot the chords
for i in range(num_chords):
    if success[i]:
        # Successful chords (orange)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='orange', alpha=0.4, lw=1)
    else:
        # Unsuccessful chords (blue)
        ax.plot([x1[i], x2[i]], [y1[i], y2[i]], color='blue', alpha=0.15, lw=1)

# Add dummy lines for the legend
ax.plot([], [], color='orange', label=r'Length $> R\sqrt{3}$')
ax.plot([], [], color='blue', label=r'Length $\leq R\sqrt{3}$')

# Formatting the plot
ax.set_aspect('equal')
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.axis('off')
ax.set_title(f"Bertrand's Paradox: Model 3 (Random Midpoint by Area)\nEmpirical Probability = {prob:.3f} (Theoretical = 0.250)")
ax.legend(loc='upper right')

plt.show()
```

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that at least one of them is a girl?

When a family has two children, there are four equally likely combinations for their genders (listing the older child first, then the younger):

Sample Space
$$\Omega = \{BB, BG, GB, GG\}$$Since each child is equally likely to be a boy or a girl, each of these 4 outcomes has a probability of $1/4$.Now,

Let's Define :

Event A : Both children are girls.
$A = \{GG\}$

Event B: At least one child is a girl.
$B = \{BG, GB, GG\}

Because we are given that at least one child is a girl, we can entirely eliminate $\{BB\}$ from our universe of possibilities.

This leaves us with only 3 possible outcomes: $\{BG, GB, GG\}$.

Out of these 3 remaining equally likely outcomes, only 1 is the event where both children are girls ($\{GG\}$).

the probability of at least one girl, is $\frac{3}{4}$.$P(A \cap B)$, the probability of both being girls and at least one being a girl, is just the probability of both being girls: $\frac{1}{4}$.$$P(A|B) = \frac{1/4}{3/4} = \frac{1}{3}$$The probability that both children are girls, given that at least one of them is a girl, is $1/3$.

```python
# Two-child problem

# Sample space: first child, second child
sample_space = ["BB", "BG", "GB", "GG"]

# Event B: at least one child is a girl
at_least_one_girl = [outcome for outcome in sample_space if "G" in outcome]

# Event A: both children are girls
both_girls = [outcome for outcome in sample_space if outcome == "GG"]

# Conditional probability P(A|B) = P(A and B) / P(B)
# Since all outcomes are equally likely, we can count outcomes

A_and_B = [outcome for outcome in at_least_one_girl if outcome in both_girls]

P_A_given_B = len(A_and_B) / len(at_least_one_girl)

print("Sample space:", sample_space)
print("At least one girl:", at_least_one_girl)
print("Both girls:", both_girls)
print("A and B:", A_and_B)
print("P(both girls | at least one girl) =", P_A_given_B)
```

#Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

###Definition of the Probability Space $(\Omega, \mathcal{F}, P)$

**1. The Sample Space ($\Omega$)**
The sample space consists of the four mutually exclusive outcomes representing the product's quality and the machine that produced it:
$$\Omega = \{G_1, G_2, D_1, D_2\}$$

**2. The $\sigma$-algebra ($\mathcal{F}$)**
The $\sigma$-algebra is given as the power set of $\Omega$, denoted as $\mathcal{P}(\Omega)$. This contains all $2^4 = 16$ possible subsets (events) of the sample space:
$$
\begin{align*}
\mathcal{F} = \{ & \emptyset, \\
& \{G_1\}, \{G_2\}, \{D_1\}, \{D_2\}, \\
& \{G_1, G_2\}, \{G_1, D_1\}, \{G_1, D_2\}, \{G_2, D_1\}, \{G_2, D_2\}, \{D_1, D_2\}, \\
& \{G_1, G_2, D_1\}, \{G_1, G_2, D_2\}, \{G_1, D_1, D_2\}, \{G_2, D_1, D_2\}, \\
& \Omega \}
\end{align*}
$$

**3. The Probability Measure ($P$)**
The probability measure $P : \mathcal{F} \to [0, 1]$ assigns a probability to any event $A \in \mathcal{F}$ defined by the sum of the probabilities of its constituent outcomes:
$$P(A) = \sum_{\omega \in A} P(\{\omega\})$$

Where the probabilities of the fundamental outcomes are given as:
* $P(\{G_1\}) = 0.50$
* $P(\{G_2\}) = 0.20$
* $P(\{D_1\}) = 0.10$
* $P(\{D_2\}) = 0.20$



#2

### Computing the $\sigma$-algebra generated by $X$: $\sigma(X)$

**1. Conceptual Definition:**
The $\sigma$-algebra generated by a discrete random variable $X$, denoted $\sigma(X)$, consists of all subsets of the sample space $\Omega$ that can be formed by the pre-images of $X$. In essence, it represents the information observable solely by knowing the value of $X$.

**2. Identifying the Partition (Pre-images):**
The random variable $X$ only takes on two values: $0$ (Good) and $1$ (Defective). We find the set of outcomes associated with each value:
* For $X = 0$: $X^{-1}(\{0\}) = \{G_1, G_2\}$
* For $X = 1$: $X^{-1}(\{1\}) = \{D_1, D_2\}$

These two sets, $\{G_1, G_2\}$ and $\{D_1, D_2\}$, form a partition of the sample space $\Omega$.

**3. Constructing $\sigma(X)$:**
To form the generated $\sigma$-algebra, we take all possible unions of these partition blocks, ensuring we include the empty set ($\emptyset$) and the entire sample space ($\Omega$).

Therefore, the $\sigma$-algebra generated by $X$ is:
$$\sigma(X) = \Big\{ \emptyset, \; \{G_1, G_2\}, \; \{D_1, D_2\}, \; \Omega \Big\}$$

#3

### Computing the $\sigma$-algebra generated by $Y$: $\sigma(Y)$

**1. Conceptual Definition:**
The $\sigma$-algebra generated by $Y$, denoted $\sigma(Y)$, consists of all subsets of $\Omega$ that can be formed by the unions of the pre-images of $Y$. Because $Y$ records *both* the quality and the machine, it provides complete information about the experiment.

**2. Identifying the Partition (Pre-images):**
The random variable $Y$ takes on four distinct values, assigning a unique number to every single outcome:
* For $Y = 1$: $Y^{-1}(\{1\}) = \{G_1\}$
* For $Y = 2$: $Y^{-1}(\{2\}) = \{G_2\}$
* For $Y = 3$: $Y^{-1}(\{3\}) = \{D_1\}$
* For $Y = 4$: $Y^{-1}(\{4\}) = \{D_2\}$

The fundamental building blocks (the partition generated by $Y$) are the individual outcomes themselves (the singletons).

**3. Constructing $\sigma(Y)$:**
To form $\sigma(Y)$, we take all possible unions of these singletons, including the empty set ($\emptyset$). Since we can form any combination of individual outcomes, $\sigma(Y)$ contains *every possible subset* of the sample space.

Therefore, the $\sigma$-algebra generated by $Y$ is the entire power set of $\Omega$, which is identical to our original $\mathcal{F}$:
$$\sigma(Y) = \mathcal{P}(\Omega) = \mathcal{F}$$

This means $\sigma(Y)$ contains all 16 possible events defined in the probability space.

### Showing $\sigma(X) \subseteq \sigma(Y)$ and Comparing Information

**1. Mathematical Proof of Inclusion:**
From our previous computations, we have the explicitly defined $\sigma$-algebras for both random variables:

$$\sigma(X) = \Big\{ \emptyset, \; \{G_1, G_2\}, \; \{D_1, D_2\}, \; \Omega \Big\}$$
$$\sigma(Y) = \mathcal{P}(\Omega)$$

To show that $\sigma(X) \subseteq \sigma(Y)$, we must verify that every element (event) in $\sigma(X)$ is also an element in $\sigma(Y)$.

Since $\sigma(Y)$ is the power set of $\Omega$, it contains *all possible subsets* of $\Omega$. By definition, the sets $\emptyset$, $\{G_1, G_2\}$, $\{D_1, D_2\}$, and $\Omega$ are all subsets of $\Omega$.

Therefore, every set in $\sigma(X)$ is strictly contained within $\sigma(Y)$, proving:
$$\sigma(X) \subseteq \sigma(Y)$$

**2. Why $Y$ Contains More Information:**
In measure theory, the "size" or granularity of a $\sigma$-algebra directly corresponds to the amount of information available.

*   **Coarse Partition:** The random variable $X$ groups outcomes together (e.g., combining $G_1$ and $G_2$ into a single measurable event). It "forgets" the machine origin, creating a coarse partition of the sample space.
*   **Fine Partition:** The random variable $Y$ assigns a unique value to every single outcome, generating a completely fine partition (singletons).

Because $\sigma(X) \subseteq \sigma(Y)$, any event that can be resolved by knowing $X$ can also be resolved by knowing $Y$. However, the reverse is not true (e.g., the event $\{G_1\}$ is in $\sigma(Y)$ but not in $\sigma(X)$). Therefore, $Y$ contains strictly more information about the underlying experiment than $X$.

### Computing the Marginal Distribution of $X$

**1. Conceptual Definition:**
The marginal distribution of a discrete random variable $X$ is a probability mass function that gives the probability of each possible value of $X$, summing over all other underlying outcomes in the sample space.

**2. Calculation:**
The random variable $X$ takes values in the set $\{0, 1\}$. To find the marginal probability for each value, we sum the probabilities of the outcomes in their respective pre-images:

*   **For $X = 0$ (Good Quality):**
    $$P(X = 0) = P(\{G_1, G_2\}) = P(\{G_1\}) + P(\{G_2\})$$
    $$P(X = 0) = 0.50 + 0.20 = 0.70$$

*   **For $X = 1$ (Defective):**
    $$P(X = 1) = P(\{D_1, D_2\}) = P(\{D_1\}) + P(\{D_2\})$$
    $$P(X = 1) = 0.10 + 0.20 = 0.30$$

**3. Final Marginal Distribution:**
The marginal distribution of $X$ is fully defined by:
*   $P(X = 0) = 0.70$
*   $P(X = 1) = 0.30$

*(Check: $\sum P(X = x) = 0.70 + 0.30 = 1.00$, confirming a valid probability distribution).*

### Computing the Marginal Distribution of $Y$

The marginal distribution of the discrete random variable $Y$ gives the probability of each distinct value $Y$ can take. ### Computing the Joint Probabilities

**1. Conceptual Definition:**
The joint probability $P(X=x, Y=y)$ is the probability of the intersection of the pre-images of $X$ and $Y$. We compute this by finding the specific outcomes $\omega \in \Omega$ that satisfy both $X(\omega) = x$ and $Y(\omega) = y$ simultaneously.

**2. Calculations:**

*   **For $P(X=0, Y=1)$:**
    The event $X=0$ corresponds to $\{G_1, G_2\}$, and $Y=1$ corresponds to $\{G_1\}$.
    $$P(X=0, Y=1) = P(\{G_1, G_2\} \cap \{G_1\}) = P(\{G_1\}) = 0.50$$

*   **For $P(X=0, Y=2)$:**
    The event $X=0$ corresponds to $\{G_1, G_2\}$, and $Y=2$ corresponds to $\{G_2\}$.
    $$P(X=0, Y=2) = P(\{G_1, G_2\} \cap \{G_2\}) = P(\{G_2\}) = 0.20$$

*   **For $P(X=1, Y=3)$:**
    The event $X=1$ corresponds to $\{D_1, D_2\}$, and $Y=3$ corresponds to $\{D_1\}$.
    $$P(X=1, Y=3) = P(\{D_1, D_2\} \cap \{D_1\}) = P(\{D_1\}) = 0.10$$

*   **For $P(X=1, Y=4)$:**
    The event $X=1$ corresponds to $\{D_1, D_2\}$, and $Y=4$ corresponds to $\{D_2\}$.
    $$P(X=1, Y=4) = P(\{D_1, D_2\} \cap \{D_2\}) = P(\{D_2\}) = 0.20$$

### Computing the Conditional Probabilities

**1. Conceptual Definition:**
The conditional probability of event $A$ given event $B$ is defined as:
$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}, \quad \text{provided } P(B) > 0$$

We will use the marginal and joint probabilities computed in the previous steps to evaluate these expressions.

**2. Calculations:**

*   **Computing $P(X=1 \mid Y=3)$:**
    $$P(X=1 \mid Y=3) = \frac{P(X=1, Y=3)}{P(Y=3)}$$
    $$P(X=1 \mid Y=3) = \frac{0.10}{0.10} = 1$$
    *(Interpretation: If $Y=3$, the outcome is strictly $D_1$. Since $D_1$ is a defective item, it guarantees $X=1$.)*

*   **Computing $P(X=1 \mid Y=4)$:**
    $$P(X=1 \mid Y=4) = \frac{P(X=1, Y=4)}{P(Y=4)}$$
    $$P(X=1 \mid Y=4) = \frac{0.20}{0.20} = 1$$
    *(Interpretation: If $Y=4$, the outcome is strictly $D_2$. This guarantees $X=1$.)*

*   **Computing $P(Y=3 \mid X=1)$:**
    $$P(Y=3 \mid X=1) = \frac{P(X=1, Y=3)}{P(X=1)}$$
    $$P(Y=3 \mid X=1) = \frac{0.10}{0.30} = \frac{1}{3} \approx 0.333$$
    *(Interpretation: Given that an item is defective, there is a $\frac{1}{3}$ probability it was produced by Machine 1.)*

*   **Computing $P(Y=4 \mid X=1)$:**
    $$P(Y=4 \mid X=1) = \frac{P(X=1, Y=4)}{P(X=1)}$$
    $$P(Y=4 \mid X=1) = \frac{0.20}{0.30} = \frac{2}{3} \approx 0.667$$
    *(Interpretation: Given that an item is defective, there is a $\frac{2}{3}$ probability it was produced by Machine 2.)*

### Interpreting the Difference Between Conditioning on $X$ and $Y$

**1. Conditioning on $X$ (Partial Information / Coarse Resolution):**
When we condition on the random variable $X$, we are updating our probability measure based on the partial information it provides (the $\sigma$-algebra $\sigma(X)$). Because $X$ maps multiple fundamental outcomes to the same value, conditioning on a specific value like $X=1$ restricts our sample space to a subset containing multiple elements (e.g., $\{D_1, D_2\}$).
* **Result:** There is still residual uncertainty. We must establish a new probability distribution *within* that subset to determine which specific machine produced the item.

**2. Conditioning on $Y$ (Complete Information / Fine Resolution):**
When we condition on the random variable $Y$, we are using the full information of the power set $\sigma(Y) = \mathcal{P}(\Omega)$. Because $Y$ provides a one-to-one mapping to the sample space, conditioning on a specific value like $Y=3$ restricts our sample space to a single deterministic outcome (the singleton $\{D_1\}$).
* **Result:** All uncertainty is resolved. The conditional probability distribution collapses into a point mass (a deterministic state) where the probability of the specific outcome becomes $1$, and the probability of all other outcomes becomes $0$.

**Conclusion:**
The fundamental difference lies in the **granularity of the $\sigma$-algebras**. Conditioning on $X$ leaves us with conditional randomness because $\sigma(X)$ is a coarse partition. Conditioning on $Y$ removes all randomness because $\sigma(Y)$ completely partitions the sample space into individual outcomes.

### Comparing the Information in $\sigma(X)$ and $\sigma(Y)$

**1. Information in $\sigma(X)$ but not in $\sigma(Y)$:**
There is **no** information available in $\sigma(X)$ that is not also available in $\sigma(Y)$. Because $\sigma(X) \subseteq \sigma(Y)$, the $\sigma$-algebra generated by $Y$ contains every measurable event present in the $\sigma$-algebra generated by $X$. Any question that can be answered by knowing $X$ can automatically be answered by knowing $Y$.

**2. Information in $\sigma(Y)$ but not in $\sigma(X)$:**
The information available in $\sigma(Y)$ but not in $\sigma(X)$ is the **machine origin** of the product.
* The $\sigma$-algebra $\sigma(X)$ only contains information capable of distinguishing the **quality** of the product (Good vs. Defective). It cannot distinguish between the fundamental outcomes $G_1$ and $G_2$, or $D_1$ and $D_2$.
* The $\sigma$-algebra $\sigma(Y)$ contains the complete granular information capable of distinguishing exactly which machine produced the item, separating the sample space into its individual fundamental outcomes. For example, the event $\{G_1\}$ (the product is Good AND came from Machine 1) is a measurable event in $\sigma(Y)$, but it is unobservable in $\sigma(X)$.
