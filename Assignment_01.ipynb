#Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius  10 . Let  A  be the event that the point lies within  1  unit of the boundary.

1.Model the experiment assuming the point is uniformly distributed with respect to area, and compute  P(A) .

2.Model the experiment assuming the distance to the point from the center is chosen uniformly from  [0,10] , and the direction is chosen independently and uniformly from  [0,2π)  and compute  P(A) .

3.Explain why the two answers differ.

# **For Part 1**

To solve this, We want to Understand some theories,

1. The Sample Space ($\Omega$): In a continuous space like a disk, $\Omega$ is the
set of all points $(x, y)$ such that their distance from the center $(0,0)$ is less than or equal to the radius $R=10$.

2. The Event ($A$): This is the "boundary region." Since the radius is $10$, being within $1$ unit of the boundary means the point's distance $r$ from the center is between $9$ and $10$.

3. Uniform Distribution (Area): When an experiment is "uniformly distributed with respect to area," the probability of an event is simply the Area of the Event divided by the Total Area of the Sample Space.

### Mathematical Modeling

**1. Sample Space ($\Omega$):**
The set of all points within the disk of radius $R = 10$:
$$\Omega = \{ (x, y) \in \mathbb{R}^2 : \sqrt{x^2 + y^2} \le 10 \}$$

**2. Event ($A$):**
The point lies within 1 unit of the boundary (between radius 9 and 10):
$$A = \{ (x, y) \in \mathbb{R}^2 : 9 \le \sqrt{x^2 + y^2} \le 10 \}$$

**3. Probability Calculation:**
Since the distribution is uniform with respect to area:
$$P(A) = \frac{\text{Area}(A)}{\text{Area}(\Omega)}$$
$$\text{Area}(\Omega) = \pi(10)^2 = 100\pi$$
$$\text{Area}(A) = \pi(10)^2 - \pi(9)^2 = 100\pi - 81\pi = 19\pi$$
$$P(A) = \frac{19\pi}{100\pi} = 0.19$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Create the figure
fig, ax = plt.subplots(figsize=(8, 8))

# Define the disks
theta = np.linspace(0, 2*np.pi, 100)
x_outer = 10 * np.cos(theta)
y_outer = 10 * np.sin(theta)
x_inner = 9 * np.cos(theta)
y_inner = 9 * np.sin(theta)

# Plot the total Sample Space (Omega)
ax.plot(x_outer, y_outer, color='black', label='Boundary (r=10)')
ax.fill(x_outer, y_outer, color='lightgray', alpha=0.3, label='Sample Space $\Omega$')

# Highlight Event A (The boundary region)
ax.fill(np.append(x_outer, x_inner[::-1]),
        np.append(y_outer, y_inner[::-1]),
        color='orange', alpha=0.6, label='Event A (Boundary Region)')

# Formatting
ax.set_aspect('equal')
ax.set_xlim(-11, 11)
ax.set_ylim(-11, 11)
ax.axhline(0, color='black', lw=1)
ax.axvline(0, color='black', lw=1)
ax.set_title("Visualization of Event A in Sample Space $\Omega$")
ax.legend()

plt.grid(True, linestyle='--')
plt.show()

print(f"Calculated Probability P(A): {19/100}")
```

**Python code to compute the probability**

```python
import numpy as np

R_outer = 10
R_inner = 9

area_full_disk = np.pi * R_outer**2
area_inner_disk = np.pi * R_inner**2
area_event_A = area_full_disk - area_inner_disk

probability_A = area_event_A / area_full_disk

print("Area of full disk =", area_full_disk)
print("Area of event A =", area_event_A)
print("P(A) =", probability_A)
```

# **For Part 02**

01.**Uniform Distance vs. Uniform Area**

The key difference here is how the probability measure is defined.

Part 1 (Uniform Area): Points are more likely to be found further from the center because there is more area at larger radius.

Part 2 (Uniform Distance/Direction): Every distance from the center ($0$ to $10$) is equally likely. This means points will appear much more crowded near the center of the disk because the same number of points are squeezed into a smaller area.




### Mathematical Modeling (Model 2: Uniform Distance and Direction)

**1. Definitions:**
Let $R$ be the random variable representing the distance from the center, and $\Theta$ be the random variable representing the direction.
- $R \sim \text{Uniform}(0, 10)$
- $\Theta \sim \text{Uniform}(0, 2\pi)$
- $R$ and $\Theta$ are independent.

**2. Event ($A$):**
The event $A$ occurs if the distance $r$ is within 1 unit of the outer boundary ($r=10$):
$$A = \{ (r, \theta) : 9 \le r \le 10 \}$$

**3. Probability Calculation:**
Because $R$ is uniformly distributed over the interval $[0, 10]$, the probability depends only on the length of the radial interval:
$$P(A) = P(9 \le R \le 10)$$
$$P(A) = \int_{9}^{10} f_R(r) \, dr = \int_{9}^{10} \frac{1}{10} \, dr$$
$$P(A) = \left[ \frac{r}{10} \right]_{9}^{10} = \frac{10}{10} - \frac{9}{10} = 0.1$$

**Conclusion:**
In this model, $P(A) = 0.1$. Note that this is lower than the area-based model ($0.19$) because this model places more probability mass near the center.

```python
import matplotlib.pyplot as plt
import numpy as np

num_points = 5000

# --- Model 1: Uniform Area ---
r_area = 10 * np.sqrt(np.random.uniform(0, 1, num_points))
theta_area = np.random.uniform(0, 2*np.pi, num_points)
x_area = r_area * np.cos(theta_area)
y_area = r_area * np.sin(theta_area)

# --- Model 2: Uniform Distance (Your current question) ---
r_dist = np.random.uniform(0, 10, num_points)
theta_dist = np.random.uniform(0, 2*np.pi, num_points)
x_dist = r_dist * np.cos(theta_dist)
y_dist = r_dist * np.sin(theta_dist)

# Plotting
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

# Plot Model 1
ax1.scatter(x_area, y_area, s=1, alpha=0.5)
circle1 = plt.Circle((0, 0), 10, color='r', fill=False, label='Boundary')
ax1.add_patch(circle1)
ax1.set_title("Model 1: Uniform Area\n$P(A) = 0.19$")
ax1.set_aspect('equal')

# Plot Model 2
ax2.scatter(x_dist, y_dist, s=1, alpha=0.5, color='green')
circle2 = plt.Circle((0, 0), 10, color='r', fill=False, label='Boundary')
ax2.add_patch(circle2)
ax2.set_title("Model 2: Uniform Distance/Direction\n$P(A) = 0.10$")
ax2.set_aspect('equal')

plt.show()
```
